<h1 style="color:#1565C0;">🐍 Program 10: Simple HTTP Server</h1>

<h3 style="color:#2E7D32;">📌 Definition</h3>

An HTTP server is a program that provides files or web resources to clients through the HTTP protocol.

<h3 style="color:#E65100;">❓ Question</h3>

Write a Python program to create a simple HTTP server for sharing project files through a local network.

<h3 style="color:#F57C00;">🎯 Aim</h3>

To create a simple HTTP server for serving files from a local directory.

<h3 style="color:#6A1B9A;">📝 Algorithm</h3>

1. Start the program.
2. Import the HTTP server module.
3. Specify the port number.
4. Create an HTTP server.
5. Start the server.
6. Serve files from the local directory.
7. Stop the server when required.
8. Stop the program.

```python
from http.server import HTTPServer, SimpleHTTPRequestHandler

PORT = 8000

server = HTTPServer(
    ("127.0.0.1", PORT),
    SimpleHTTPRequestHandler
)

print("HTTP Server running at http://127.0.0.1:8000")

try:
    server.serve_forever()

except KeyboardInterrupt:
    print("\nServer stopped.")

finally:
    server.server_close()
```
