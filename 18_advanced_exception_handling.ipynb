<h1 style="color:#1565C0;">🐍 Program 18: Advanced Exception Handling</h1>

<h3 style="color:#2E7D32;">📌 Definition</h3>

Advanced exception handling uses multiple exceptions, else, finally and custom exceptions to manage different types of errors.

<h3 style="color:#E65100;">❓ Question</h3>

Write a Python program to demonstrate advanced exception handling in an online banking transaction.

<h3 style="color:#F57C00;">🎯 Aim</h3>

To demonstrate advanced exception handling using multiple exceptions, else, finally and custom exceptions.

<h3 style="color:#6A1B9A;">📝 Algorithm</h3>

1. Start the program.
2. Define a custom exception.
3. Set the account balance.
4. Get the withdrawal amount.
5. Check for invalid amount.
6. Check for insufficient balance.
7. Handle exceptions.
8. Execute the else block for successful transactions.
9. Execute the finally block.
10. Stop the program.

```python
class InsufficientBalanceError(Exception):
    pass


balance = 5000

try:
    amount = float(input("Enter withdrawal amount: "))

    if amount <= 0:
        raise ValueError("Amount must be positive.")

    if amount > balance:
        raise InsufficientBalanceError(
            "Insufficient balance."
        )

except ValueError as e:
    print("Value Error:", e)

except InsufficientBalanceError as e:
    print("Transaction Error:", e)

except Exception as e:
    print("Unexpected Error:", e)

else:
    balance -= amount

    print("Transaction successful.")
    print("Remaining balance:", balance)

finally:
    print("Transaction process completed.")
```
