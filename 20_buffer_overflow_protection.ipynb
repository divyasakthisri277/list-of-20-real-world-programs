<h1 style="color:#1565C0;">🐍 Program 20: Buffer Overflow Protection</h1>

<h3 style="color:#2E7D32;">📌 Definition</h3>

Buffer overflow protection prevents input from exceeding the allowed storage size by using input validation and bounds checking.

<h3 style="color:#E65100;">❓ Question</h3>

Write a Python program to demonstrate buffer overflow protection using input validation and safe parsing.

<h3 style="color:#F57C00;">🎯 Aim</h3>

To demonstrate buffer overflow protection by validating input length and safely processing input.

<h3 style="color:#6A1B9A;">📝 Algorithm</h3>

1. Start the program.
2. Set the maximum allowed input length.
3. Get the customer account number.
4. Check the input length.
5. Validate that the input contains only digits.
6. Accept valid input.
7. Reject invalid or oversized input.
8. Stop the program.

```python
MAX_LENGTH = 20

customer_input = input(
    "Enter customer account number: "
)

if len(customer_input) > MAX_LENGTH:

    print("Input rejected: length exceeds limit.")

elif not customer_input.isdigit():

    print("Input rejected: invalid input.")

else:

    print("Input accepted safely.")
    print("Account number:", customer_input)
```
