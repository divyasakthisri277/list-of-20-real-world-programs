<h1 style="color:#1565C0;">🐍 Program 7: Slicing Lists</h1>

<h3 style="color:#2E7D32;">📌 Definition</h3>

List slicing is used to access a specific portion of a list using start, stop and step values.

<h3 style="color:#E65100;">❓ Question</h3>

Write a Python program to analyze student marks using different list slicing techniques.

<h3 style="color:#F57C00;">🎯 Aim</h3>

To demonstrate positive slicing, negative slicing, step slicing and reversing a list.

<h3 style="color:#6A1B9A;">📝 Algorithm</h3>

1. Start the program.
2. Create a list of student marks.
3. Apply positive slicing.
4. Apply negative slicing.
5. Apply step slicing.
6. Reverse the list.
7. Display all results.
8. Stop the program.

```python
marks = [45, 67, 78, 89, 92, 56, 73, 81]

print("Original List:", marks)

print("Positive Slicing:", marks[1:5])

print("Negative Slicing:", marks[-6:-2])

print("Step Slicing:", marks[::2])

print("Reversed List:", marks[::-1])
```
