<h1 style="color:#1565C0;">🐍 Program 16: Socket Server Framework</h1>

<h3 style="color:#2E7D32;">📌 Definition</h3>

A socket server is a program that listens for client connections and communicates with connected clients.

<h3 style="color:#E65100;">❓ Question</h3>

Write a Python program to create a simple TCP socket server for a college chat application.

<h3 style="color:#F57C00;">🎯 Aim</h3>

To implement a simple TCP socket server for client communication.

<h3 style="color:#6A1B9A;">📝 Algorithm</h3>

1. Start the server.
2. Create a TCP socket.
3. Bind the socket to an IP address and port.
4. Listen for a client connection.
5. Accept the connection.
6. Receive the client message.
7. Send a response.
8. Close the connection.
9. Stop the server.

```python
import socket

server = socket.socket(
    socket.AF_INET,
    socket.SOCK_STREAM
)

server.bind(("127.0.0.1", 6000))

server.listen(1)

print("Server is waiting for connection...")

connection, address = server.accept()

print("Connected with:", address)

data = connection.recv(1024)

if data:
    print("Client:", data.decode())

    response = "Message received"
    connection.send(response.encode())

connection.close()
server.close()

print("Server stopped.")
```
