<h1 style="color:#1565C0;">🐍 Program 12: Ping Sweep</h1>

<h3 style="color:#2E7D32;">📌 Definition</h3>

A ping sweep checks a range of IP addresses to identify reachable hosts in an authorized network.

<h3 style="color:#E65100;">❓ Question</h3>

Write a Python program to perform a ping sweep on a specified local IP range.

<h3 style="color:#F57C00;">🎯 Aim</h3>

To identify reachable hosts in an authorized local network.

<h3 style="color:#6A1B9A;">📝 Algorithm</h3>

1. Start the program.
2. Enter the authorized local network prefix.
3. Generate IP addresses.
4. Send a ping request to each address.
5. Check the response.
6. Display reachable hosts.
7. Stop the program.

```python
import subprocess
import platform

network = input(
    "Enter local network prefix (example: 127.0.0.): "
)

for i in range(1, 11):

    ip = network + str(i)

    if platform.system().lower() == "windows":
        command = ["ping", "-n", "1", "-w", "500", ip]
    else:
        command = ["ping", "-c", "1", "-W", "1", ip]

    result = subprocess.run(
        command,
        stdout=subprocess.DEVNULL,
        stderr=subprocess.DEVNULL
    )

    if result.returncode == 0:
        print(ip, "is reachable")
```
