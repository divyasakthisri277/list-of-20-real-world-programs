<h1 style="color:#1565C0;">🐍 Program 17: HTTP Client with Sockets</h1>

<h3 style="color:#2E7D32;">📌 Definition</h3>

An HTTP client communicates with a web server by establishing a connection and sending HTTP requests.

<h3 style="color:#E65100;">❓ Question</h3>

Write a Python program using sockets to connect to an authorized HTTP server, send a GET request and display the response.

<h3 style="color:#F57C00;">🎯 Aim</h3>

To understand HTTP communication using a socket and a GET request.

<h3 style="color:#6A1B9A;">📝 Algorithm</h3>

1. Start the program.
2. Create a TCP socket.
3. Specify the HTTP server.
4. Connect to the server.
5. Create an HTTP GET request.
6. Send the request.
7. Receive the response.
8. Display the response.
9. Close the socket.
10. Stop the program.

```python
import socket

host = "example.com"
port = 80

client = socket.socket(
    socket.AF_INET,
    socket.SOCK_STREAM
)

try:
    client.settimeout(10)

    client.connect((host, port))

    request = (
        "GET / HTTP/1.1\r\n"
        "Host: example.com\r\n"
        "Connection: close\r\n"
        "\r\n"
    )

    client.sendall(request.encode())

    response = b""

    while True:
        data = client.recv(4096)

        if not data:
            break

        response += data

    print(response.decode(errors="ignore"))

except socket.error as e:
    print("Connection error:", e)

finally:
    client.close()
```
