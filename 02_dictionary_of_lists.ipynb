<h1 style="color:#1565C0;">🐍 Program 2: Dictionary of Lists</h1>

<h3 style="color:#2E7D32;">📌 Definition</h3>

A dictionary of lists stores multiple values as a list for each dictionary key.

<h3 style="color:#E65100;">❓ Question</h3>

Write a Python program to manage students department-wise using a dictionary of lists.

<h3 style="color:#F57C00;">🎯 Aim</h3>

To store, add and search student names department-wise using a dictionary of lists.

<h3 style="color:#6A1B9A;">📝 Algorithm</h3>

1. Start the program.
2. Create a dictionary containing departments and students.
3. Get the department and student name.
4. Add the student to the selected department.
5. Search for a student name.
6. Display the department-wise student list.
7. Stop the program.

```python
departments = {
    "Computer Science": ["Asha", "Priya"],
    "Cyber Security": ["Merlin","divya"],
    "Mathematics": ["Anu"]
}

department = input("Enter department: ")
name = input("Enter student name: ")

if department in departments:
    departments[department].append(name)
    print("Student added successfully.")
else:
    departments[department] = [name]
    print("New department added successfully.")

search_name = input("Enter student name to search: ")

found = False

for dept, students in departments.items():
    if search_name in students:
        print(search_name, "found in", dept)
        found = True

if not found:
    print(search_name, "not found.")

print("\nDepartment-wise Students:")

for dept, students in departments.items():
    print(dept, ":", students)
```
