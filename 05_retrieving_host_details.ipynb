<h1 style="color:#1565C0;">🐍 Program 5: Retrieving Host Details</h1>

<h3 style="color:#2E7D32;">📌 Definition</h3>

Host details include the computer's host name and IP address, which help identify a computer on a network.

<h3 style="color:#E65100;">❓ Question</h3>

Write a Python program to retrieve and display the host name and IP address of a computer.

<h3 style="color:#F57C00;">🎯 Aim</h3>

To retrieve and display the host name and IP address of a computer.

<h3 style="color:#6A1B9A;">📝 Algorithm</h3>

1. Start the program.
2. Import the socket module.
3. Retrieve the host name.
4. Retrieve the IP address.
5. Display the host name and IP address.
6. Stop the program.

```python
import socket

hostname = socket.gethostname()

try:
    ip_address = socket.gethostbyname(hostname)

    print("Host Name:", hostname)
    print("IP Address:", ip_address)

except socket.error as e:
    print("Error:", e)
```
