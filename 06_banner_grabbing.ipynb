<h1 style="color:#1565C0;">🐍 Program 6: Banner Grabbing</h1>

<h3 style="color:#2E7D32;">📌 Definition</h3>

Banner grabbing is the process of retrieving service information from an authorized server.

<h3 style="color:#E65100;">❓ Question</h3>

Write a Python program to connect to an authorized server and retrieve its service banner.

<h3 style="color:#F57C00;">🎯 Aim</h3>

To connect to an authorized server and retrieve its service banner.

<h3 style="color:#6A1B9A;">📝 Algorithm</h3>

1. Start the program.
2. Enter the authorized server IP address and port.
3. Create a TCP socket.
4. Connect to the server.
5. Receive the service banner.
6. Display the banner.
7. Close the socket.
8. Stop the program.

```python
import socket

host = input("Enter authorized server IP: ")
port = int(input("Enter port number: "))

try:
    sock = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
    sock.settimeout(5)

    sock.connect((host, port))

    banner = sock.recv(1024)

    print("Service Banner:")
    print(banner.decode(errors="ignore"))

    sock.close()

except ConnectionRefusedError:
    print("Connection refused. Make sure the authorized server is running.")

except socket.timeout:
    print("Connection timed out.")

except Exception as e:
    print("Connection error:", e)
```
