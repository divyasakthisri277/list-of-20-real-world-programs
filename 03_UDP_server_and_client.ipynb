<h1 style="color:#1565C0;">🐍 Program 3: UDP Server and Client</h1>

<h3 style="color:#2E7D32;">📌 Definition</h3>

UDP is a connectionless communication protocol used to send and receive data through datagram packets.

<h3 style="color:#E65100;">❓ Question</h3>

Write a Python program using UDP server and client to send and receive messages between two computers.

<h3 style="color:#F57C00;">🎯 Aim</h3>

To implement a simple messaging system using UDP server and client.

<h3 style="color:#6A1B9A;">📝 Algorithm – UDP Server</h3>

1. Start the server.
2. Create a UDP socket.
3. Bind the socket to an IP address and port.
4. Wait for a message from the client.
5. Receive and display the message.
6. Send a response to the client.
7. Close the socket.
8. Stop the program.

<h3 style="color:#6A1B9A;">📝 Algorithm – UDP Client</h3>

1. Start the client.
2. Create a UDP socket.
3. Enter a message.
4. Send the message to the server.
5. Receive the server response.
6. Display the response.
7. Close the socket.
8. Stop the program.

```python
import socket

server = socket.socket(socket.AF_INET, socket.SOCK_DGRAM)

server.bind(("127.0.0.1", 5000))

print("UDP Server is waiting...")

data, address = server.recvfrom(1024)

print("Client:", data.decode())

response = "Message received successfully"

server.sendto(response.encode(), address)

server.close()
```

```python
import socket

client = socket.socket(socket.AF_INET, socket.SOCK_DGRAM)

message = input("Enter message: ")

client.sendto(message.encode(), ("127.0.0.1", 5000))

data, address = client.recvfrom(1024)

print("Server:", data.decode())

client.close()
```
