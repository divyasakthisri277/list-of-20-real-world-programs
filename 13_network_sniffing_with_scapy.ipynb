<h1 style="color:#1565C0;">🐍 Program 13: Network Sniffing with Scapy</h1>

<h3 style="color:#2E7D32;">📌 Definition</h3>

Network sniffing is the process of capturing and examining network packets for monitoring and analysis.

<h3 style="color:#E65100;">❓ Question</h3>

Write a Python program using Scapy to capture packets in an authorized laboratory network and display their source, destination and protocol.

<h3 style="color:#F57C00;">🎯 Aim</h3>

To capture packets in an authorized laboratory network and display packet information.

<h3 style="color:#6A1B9A;">📝 Algorithm</h3>

1. Start the program.
2. Import the Scapy library.
3. Start packet capture.
4. Capture packets.
5. Extract source address.
6. Extract destination address.
7. Identify the protocol.
8. Display packet information.
9. Stop packet capture.
10. Stop the program.

```python
from scapy.all import sniff, IP

def display_packet(packet):

    if IP in packet:

        source = packet[IP].src
        destination = packet[IP].dst
        protocol = packet[IP].proto

        print("Source:", source)
        print("Destination:", destination)
        print("Protocol:", protocol)
        print("-" * 40)

print("Capturing packets...")
print("Use only on an authorized laboratory network.")

sniff(
    prn=display_packet,
    count=10
)

print("Packet capture completed.")
```
