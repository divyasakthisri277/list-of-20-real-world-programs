<h1 style="color:#1565C0;">🐍 Program 8: Exception Handling in File Operations</h1>

<h3 style="color:#2E7D32;">📌 Definition</h3>

Exception handling in file operations is used to safely manage errors while creating, writing and reading files.

<h3 style="color:#E65100;">❓ Question</h3>

Write a Python program to perform file operations and handle file-related exceptions.

<h3 style="color:#F57C00;">🎯 Aim</h3>

To perform file operations safely using exception handling.

<h3 style="color:#6A1B9A;">📝 Algorithm</h3>

1. Start the program.
2. Get student information.
3. Open the file in write mode.
4. Write the information into the file.
5. Open the file in read mode.
6. Display the file content.
7. Handle file-related exceptions.
8. Stop the program.

```python
try:
    student_info = input("Enter student information: ")
    additional_info = input("Enter additional information: ")

    with open("student.txt", "w") as file:
        file.write(student_info + "\n")
        file.write(additional_info + "\n")

    with open("student.txt", "r") as file:
        print("\nFile Content:")
        print(file.read())

except FileNotFoundError:
    print("File not found.")

except PermissionError:
    print("Permission denied.")

except Exception as e:
    print("Error:", e)
```
