<h1 style="color:#1565C0;">🐍 Program 9: Using Global Variables in Functions</h1>

<h3 style="color:#2E7D32;">📌 Definition</h3>

A global variable is a variable declared outside a function and can be accessed or modified inside a function using the global keyword.

<h3 style="color:#E65100;">❓ Question</h3>

Write a Python program using a global variable to manage an online shopping cart amount.

<h3 style="color:#F57C00;">🎯 Aim</h3>

To use a global variable in functions to display and modify a shopping cart amount.

<h3 style="color:#6A1B9A;">📝 Algorithm</h3>

1. Start the program.
2. Create a global variable for cart amount.
3. Define a function to display the cart amount.
4. Define a function to add an item amount.
5. Use the global keyword to modify the cart amount.
6. Display the updated amount.
7. Stop the program.

```python
cart_amount = 0

def display_cart():
    print("Cart Amount:", cart_amount)

def add_item(amount):
    global cart_amount
    cart_amount += amount

display_cart()

amount = float(input("Enter item price: "))

add_item(amount)

display_cart()
```
