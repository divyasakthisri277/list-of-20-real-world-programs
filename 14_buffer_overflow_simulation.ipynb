<h1 style="color:#1565C0;">🐍 Program 14: Buffer Overflow Simulation</h1>

<h3 style="color:#2E7D32;">📌 Definition</h3>

A buffer overflow occurs when input exceeds the available storage capacity. This program safely simulates the condition by checking input size.

<h3 style="color:#E65100;">❓ Question</h3>

Write a Python program to safely simulate a buffer overflow condition by checking input size.

<h3 style="color:#F57C00;">🎯 Aim</h3>

To safely simulate a buffer overflow condition by checking the size of user input.

<h3 style="color:#6A1B9A;">📝 Algorithm</h3>

1. Start the program.
2. Define the maximum input size.
3. Get input from the user.
4. Calculate the input length.
5. Compare the input length with the maximum size.
6. Accept safe input.
7. Display a warning for oversized input.
8. Stop the program.

```python
MAX_SIZE = 10

data = input("Enter input: ")

print("Input length:", len(data))

if len(data) <= MAX_SIZE:
    print("Input accepted safely.")
else:
    print("Warning: Input size exceeds the allowed limit.")
```
