<h3 style="color:#2E7D32;">📌 Definition</h3>

A SYN flood is a denial-of-service concept involving many connection requests. This program safely simulates the concept without sending real attack traffic.

<h3 style="color:#E65100;">❓ Question</h3>

Write a Python program to safely simulate the concept of excessive SYN requests in a local laboratory environment.

<h3 style="color:#F57C00;">🎯 Aim</h3>

To safely simulate the concept of excessive SYN requests in a laboratory environment.

<h3 style="color:#6A1B9A;">📝 Algorithm</h3>

1. Start the program.
2. Define the maximum connection limit.
3. Generate simulated connection requests.
4. Count the requests.
5. Compare the count with the connection limit.
6. Display the simulated effect.
7. Clear the simulated data.
8. Stop the program.

```python
MAX_CONNECTIONS = 5

requests = []

for i in range(10):
    requests.append("SYN Request " + str(i + 1))

print("Total simulated requests:", len(requests))
print("Maximum connection limit:", MAX_CONNECTIONS)

if len(requests) > MAX_CONNECTIONS:
    print("Simulation: Connection limit exceeded.")
    print("Server may be unable to handle additional requests.")
else:
    print("Simulation: Server can handle the requests.")

requests.clear()
```
