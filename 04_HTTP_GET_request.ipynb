<h1 style="color:#1565C0;">🐍 Program 4: HTTP GET Request</h1>

<h3 style="color:#2E7D32;">📌 Definition</h3>

HTTP GET is a request method used to retrieve information from a web server.

<h3 style="color:#E65100;">❓ Question</h3>

Write a Python program to send an HTTP GET request and display the status code, headers and response content.

<h3 style="color:#F57C00;">🎯 Aim</h3>

To send an HTTP GET request and display the response details.

<h3 style="color:#6A1B9A;">📝 Algorithm</h3>

1. Start the program.
2. Import the requests module.
3. Specify the website URL.
4. Send an HTTP GET request.
5. Receive the response.
6. Display the status code.
7. Display the headers.
8. Display the response content.
9. Handle request errors.
10. Stop the program.

```python
import requests

url = "https://example.com"

try:
    response = requests.get(url, timeout=10)

    print("Status Code:", response.status_code)

    print("\nHeaders:")
    print(response.headers)

    print("\nResponse Content:")
    print(response.text[:500])

except requests.RequestException as e:
    print("Request error:", e)
```
