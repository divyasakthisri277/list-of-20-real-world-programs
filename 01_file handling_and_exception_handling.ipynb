<h1 style="color:#1565C0;">🐍 Program 1: File Handling and Exception Handling</h1>

<h2 style="color:#2E7D32;">📌 Definition</h2>

File handling is used to create, read, write and append data in files.
Exception handling is used to handle errors that may occur during file operations.

<h2 style="color:#D84315;">❓ Question</h2>

A college wants to maintain student records in a text file.
Write a Python program to add, read and append student details
and handle file-related errors.

<h2 style="color:#E65100;">🎯 Aim</h2>

To write a Python program to add, read and append student details
in a text file and handle file-related errors.

<h2 style="color:#6A1B9A;">📝 Algorithm</h2>

1. Start the program.
2. Get the student details from the user.
3. Open a text file in append mode.
4. Write the student details into the file.
5. Open the file in read mode.
6. Read and display the stored student details.
7. Handle file-related errors using exception handling.
8. Close the file.
9. Stop the program.

<h2 style="color:#00838F;">💻 Program</h2>

```python
try:
    name = input("Enter student name: ")
    roll_no = input("Enter roll number: ")
    department = input("Enter department: ")

    with open("students.txt", "w") as file:
        file.write(f"{roll_no}, {name}, {department}\n")

    print("Student record added successfully.")

    with open("students.txt", "r") as file:
        print("\nStudent Records:")
        print(file.read())

except FileNotFoundError:
    print("File not found.")

except Exception as e:
    print("An error occurred:", e)
```
